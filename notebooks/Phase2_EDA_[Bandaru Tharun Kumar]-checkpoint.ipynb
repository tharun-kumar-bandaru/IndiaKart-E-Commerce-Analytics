# Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
output_folder = "outputs/phase2_charts"
os.makedirs(output_folder, exist_ok=True)
```

# Loading all 8 CSV files

```python
customers = pd.read_csv("customers DA.csv")
inventory = pd.read_csv("inventory DA.csv")
order_items = pd.read_csv("order_items DA.csv")
orders = pd.read_csv("orders DA.csv")
payments = pd.read_csv("payments DA.csv")
products = pd.read_csv("products DA.csv")
returns = pd.read_csv("returns DA.csv")
suppliers = pd.read_csv("suppliers DA.csv")
```

```python
print("Customers:", customers.shape)
print("Inventory:", inventory.shape)
print("Order Items:", order_items.shape)
print("Orders:", orders.shape)
print("Payments:", payments.shape)
print("Products:", products.shape)
print("Returns:", returns.shape)
print("Suppliers:", suppliers.shape)
```

# Prepare the Data

```python
orders["order_date"] = pd.to_datetime(
    orders["order_date"],
    dayfirst=True,
    errors="coerce"
)
```

```python
print(orders["order_date"].min())
print(orders["order_date"].max())
```

# Convert amount columns

```python
orders["final_amount"] = pd.to_numeric(
    orders["final_amount"],
    errors="coerce"
)

orders["gst_amount"] = pd.to_numeric(
    orders["gst_amount"],
    errors="coerce"
)

order_items["total_price"] = pd.to_numeric(
    order_items["total_price"],
    errors="coerce"
)

order_items["quantity"] = pd.to_numeric(
    order_items["quantity"],
    errors="coerce"
)

customers["age"] = pd.to_numeric(
    customers["age"],
    errors="coerce"
)
```

# Create Month and Year Columns

```python
orders["year"] = orders["order_date"].dt.year

orders["month"] = orders["order_date"].dt.month

orders["month_name"] = orders["order_date"].dt.strftime("%b")

orders["year_month"] = orders["order_date"].dt.to_period("M")
```

```python
orders[
    ["order_date", "year", "month", "month_name", "year_month"]
].head()
```

# CHART 1 — Monthly Order Volume

#Requirement
Monthly order volume — bar or line chart showing number of orders per month for all 24 months. Highlight October and November.

### Calculate

```python
monthly_orders = (
    orders
    .groupby("year_month")["order_id"]
    .nunique()
    .reset_index(name="order_count")
)

monthly_orders["year_month"] = monthly_orders["year_month"].astype(str)

monthly_orders.head()
```

### Create chart

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_orders["year_month"],
    monthly_orders["order_count"],
    marker="o"
)

plt.title("Monthly Order Volume")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    f"{output_folder}/01_monthly_order_volume.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Find highest and lowest months

```python
highest_order_month = monthly_orders.loc[
    monthly_orders["order_count"].idxmax()
]

lowest_order_month = monthly_orders.loc[
    monthly_orders["order_count"].idxmin()
]

print("Highest order month:")
print(highest_order_month)

print("\nLowest order month:")
print(lowest_order_month)
```

### Check October and November

```python
highest_order_month = monthly_orders.loc[
    monthly_orders["order_count"].idxmax()
]

lowest_order_month = monthly_orders.loc[
    monthly_orders["order_count"].idxmin()
]

print("Highest order month:")
print(highest_order_month)

print("\nLowest order month:")
print(lowest_order_month)
```

### Observation
Order volume shows noticeable variation across the 24-month period, with certain months recording significantly higher order activity than others. October and November should be examined closely because they may indicate seasonal or promotional demand patterns.

# CHART 2 — Monthly Revenue / GMV

### Requirement
Monthly revenue using final_amount. Compare year-over-year.

```python
monthly_revenue = (
    orders
    .groupby("year_month")["final_amount"]
    .sum()
    .reset_index(name="revenue")
)

monthly_revenue["year_month"] = (
    monthly_revenue["year_month"].astype(str)
)

monthly_revenue.head()
```

```python
#Chart
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_revenue["year_month"],
    monthly_revenue["revenue"],
    marker="o"
)

plt.title("Monthly Revenue (GMV)")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    f"{output_folder}/02_monthly_revenue.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Yearly revenue
yearly_revenue = (
    orders
    .groupby("year")["final_amount"]
    .sum()
    .reset_index()
)

yearly_revenue
```

```python
#YoY calculation
yearly_revenue["YoY_Growth_%"] = (
    yearly_revenue["final_amount"]
    .pct_change() * 100
)

yearly_revenue
```

### Observation format
Total GMV shows a [rising/falling/fluctuating] trend over the analysis period. Year-over-year revenue changed by approximately 88.28%, indicating [growth/decline] in overall business performance. The strongest revenue month was -51.35, which may be associated with increased order volume, seasonal demand or promotional activity.

# CHART 3 — Category-wise Revenue Share

### Requirement
Which category earns the most?

```python
category_revenue = (
    order_items
    .groupby("category")["total_price"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

category_revenue
```

```python
#Calculate percentage
category_revenue["revenue_share_%"] = (
    category_revenue["total_price"]
    / category_revenue["total_price"].sum()
    * 100
)

category_revenue
```

```python
#Horizontal bar chart
plt.figure(figsize=(10, 6))

plt.barh(
    category_revenue["category"],
    category_revenue["total_price"]
)

plt.title("Category-wise Revenue")
plt.xlabel("Revenue (₹)")
plt.ylabel("Category")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.savefig(
    f"{output_folder}/03_category_revenue.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Find top category
top_category = category_revenue.iloc[0]

print("Top Revenue Category:", top_category["category"])
print("Revenue Share:", round(top_category["revenue_share_%"], 2), "%")
```

### Observation
The highest-revenue category is Electronics 57.41%, contributing approximately 0.49% of total category revenue. The lowest-performing category is , suggesting that management should investigate differences in demand, pricing, product availability and customer preferences across categories.

# CHART 10 — Average Order Value by Customer Segment


### Requirement
Bar chart. Do Premium customers spend more per order?

```python
orders_customer = orders.merge(
    customers[
        ["customer_id", "segment"]
    ],
    on="customer_id",
    how="left"
)

orders_customer.head()
```

```python
#AOV = Total Revenue / Number of Orders
aov_segment = (
    orders_customer
    .groupby("segment")
    .agg(
        total_revenue=("final_amount", "sum"),
        total_orders=("order_id", "nunique")
    )
    .reset_index()
)

aov_segment["AOV"] = (
    aov_segment["total_revenue"]
    / aov_segment["total_orders"]
)

aov_segment
```

```python
#Chart
plt.figure(figsize=(10, 6))

plt.bar(
    aov_segment["segment"],
    aov_segment["AOV"]
)

plt.title("Average Order Value by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Average Order Value (₹)")

plt.tight_layout()

plt.savefig(
    f"{output_folder}/10_aov_by_segment.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Find highest AOV
highest_aov_segment = aov_segment.loc[
    aov_segment["AOV"].idxmax()
]

print(
    "Highest AOV Segment:",
    highest_aov_segment["segment"]
)

print(
    "AOV:",
    round(highest_aov_segment["AOV"], 2)
)
```

### Observation
The segment has the highest average order value at approximately ₹₹64,124.56per order. If Premium customers have the highest AOV, this supports the idea that Premium customers are particularly valuable and may justify personalized offers, loyalty benefits and targeted recommendations.
